```python
# TODO 05 - how much total sales, average sales, and standard deviation of sales your company make in 2017
print(df['Sales'].sum(), df['Sales'].mean(), df['Sales'].std())
```

```python
# TODO 06 - which Segment has the highest profit in 2018
df_2018 = df[df['Order Date'].dt.year == 2018]
profits = df_2018.groupby('Segment')['Profit'].sum()
profits
```

```python
# TODO 07 - which top 5 States have the least total sales between 15 April 2019 - 31 December 2019
df_date = df[(df['Order Date'] >= pd.Timestamp('2019-04-15')) & (df['Order Date'] <= pd.Timestamp('2019-12-31'))]
least = df_date.groupby('State')['Sales'].sum().nsmallest(5)
least
```

```python
# TODO 08 - what is the proportion of total sales (%) in West + Central in 2019 e.g. 25% 
west_df = df[((df['Order Date'].dt.year == 2019) & (df['Region'] == 'West'))]
central_df = df[((df['Order Date'].dt.year == 2019) & (df['Region'] == 'Central'))]
total_df = df[df['Order Date'].dt.year == 2019]
# obtain total sales in West & Central regions in 2019
west_sales = west_df['Sales'].sum()
central_sales = central_df['Sales'].sum()
# obtain total sales in 2019
total_sales = total_df['Sales'].sum()

# compute proportion of sales
sales_proportion_west = (west_sales / total_sales) * 100
sales_proportion_central = (central_sales / total_sales) * 100
print('Sales in West = %.1f %%' % sales_proportion_west,
      'and sales in Central = %.1f %%' % sales_proportion_central)
```

```python
# TODO 09 - find top 10 popular products in terms of number of orders vs. total sales during 2019-2020
popular_products = df.groupby('Product Name').agg({'Sales':'sum'}).nlargest(10, 'Sales')
popular_products
```
